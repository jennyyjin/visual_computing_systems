```python
import time

start = time.time()

!PYTORCH_CUDA_ALLOC_CONF=expandable_segments:True \
 ./.venv/bin/python -m ltx_pipelines.distilled \
  --distilled-checkpoint-path checkpoints/LTX-2.3/ltx-2.3-22b-distilled-1.1.safetensors \
  --spatial-upsampler-path checkpoints/LTX-2.3/ltx-2.3-spatial-upscaler-x2-1.1.safetensors \
  --gemma-root checkpoints/LTX-2.3/gemma \
  --prompt "test prompt" \
  --num_inference_steps 4 \
  --height 384 \
  --width 640 \
  --num-frames 49 \
  --frame-rate 24 \
  --output-path ./output.mp4 \
  --quantization fp8-cast

end = time.time()
print("Total time:", end - start)
```
