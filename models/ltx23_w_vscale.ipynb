```python
%%bash
cd /root/LTX-2

mkdir -p packages/ltx-pipelines/src/ltx_pipelines/multigpu

cat > packages/ltx-pipelines/src/ltx_pipelines/multigpu/__init__.py <<'EOF'
EOF

cat > packages/ltx-pipelines/src/ltx_pipelines/multigpu/delegating_builder.py <<'EOF'
class DelegatingBuilder:
    def __init__(self, *args, **kwargs):
        self.args = args
        self.kwargs = kwargs

    def __enter__(self):
        return self

    def __exit__(self, *args):
        return False
EOF
```

```python
from huggingface_hub import list_repo_files

files = list_repo_files("Lightricks/LTX-2")

for f in files:
    if any(x in f.lower() for x in [
        "safetensors",
        "distilled",
        "upscaler",
        "2.3"
    ]):
        print(f)
```

```python
%%bash
cd /root/LTX-2
mkdir -p checkpoints/LTX-2.3

hf download Lightricks/LTX-2 \
  ltx-2-19b-distilled.safetensors \
  --local-dir checkpoints/LTX-2.3

hf download Lightricks/LTX-2 \
  ltx-2-spatial-upscaler-x2-1.0.safetensors \
  --local-dir checkpoints/LTX-2.3
```

```python
import time

start = time.time()

!PYTORCH_ALLOC_CONF=expandable_segments:True \
 ./.venv/bin/python -m ltx_pipelines.distilled \
  --distilled-checkpoint-path checkpoints/LTX-2.3/ltx-2-19b-distilled.safetensors \
  --spatial-upsampler-path checkpoints/LTX-2.3/ltx-2-spatial-upscaler-x2-1.0.safetensors \
  --gemma-root checkpoints/LTX-2.3/gemma \
  --prompt "A realistic magician in a black tuxedo standing on a theater stage dramatically pulls a fluffy white rabbit out of a black magic hat, cinematic lighting, detailed face, realistic motion" \
  --height 512 \
  --width 512 \
  --num-frames 49 \
  --frame-rate 24 \
  --output-path ./output.mp4 \
  --quantization fp8-cast

end = time.time()
print("Total time:", end - start)
```
